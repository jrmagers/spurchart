## Upconverting Transmitter

Example Transmit: 3 GHz upconvert to 12 GHz

```python
tuc = transmit(fif=3, frf=(0, 25), flo=(10,20), bw=0.8)
tuc.band(1,-1, (11,13), color='blue')
tuc.band(-1,1, (19,21), color='green')
tuc
# tuc.describe()
```

## Upconverting Receiver

![Downconverting Receiver](../mixer/RxUp.svg)

A good example of an upconverting receiver is the "lowband" signal chain of a spectrum analyzer tuned.
the input to the mixer is the IF port of a double balanced mixer 
the purpose of the lowpass filter is to reject the RF frequency of 4730 MHz, provide image rejection (-1,1), and reject the LO leaking through the IF port of the mixer to the front-panel of the spectrum analyzer.

```python
ruc = receive(frf=4730, flo=(4730,4730+3600), bw=320, fif=(0, 5000), units='MHz', threshold=-110)
ruc.band(1,-1, (0,3600), color='purple')
ruc
```

```python
receive(frf=4730, flo=6500, bw=320, fif=(0, 5000), units='MHz', threshold=-110)
```
