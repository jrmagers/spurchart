```python
from spurchart.converter import Conversion
```

## DAC Example
- DAC39RF20 Running at 38.68 GS/s

```python
fs = 2*19.84  # sample rate
```

Jesse set DAC39JRF20 to a sample rate of 2 x 19.84 MS/s


```python
dac = Conversion(
    fs=fs,
    order=(4, 2),
    M=16,
    units="GHz",
    input_zone=1,
    tune_zone=1,
    deduplicate=True,
    converter = 'dac'
)
```

```python
f1 = 13.004995759
spurs = dac.spurs_at_fc(f1)

spurs = spurs[(spurs.index >= 13) & (spurs.index <=17)]
spurs['relative to 15 GHz'] = spurs.index - 15
spurs
```

```python
dac.addband(fa=13, fb=17, color="blue")

dac.plot(legend=False, swap_axes=False)
```

```python
n = -2
k = 0
fc = f1
offset = abs(n*fc + k*fs/16)-fs/2
fs/2-offset
```
